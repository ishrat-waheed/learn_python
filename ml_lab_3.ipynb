```python
import pandas as pd
from sklearn.datasets import load_iris # Load the Iris dataset
data = load_iris()
df = pd.DataFrame(data.data, columns=data.feature_names)
print(df.head())
```

```python
from sklearn.preprocessing import MinMaxScaler # Initialize the scaler scaler = MinMaxScaler() # Fit and transform the data df_minmax = pd.DataFrame(scaler.fit_transform(df), columns=df.columns) print(df_minmax.head())
```

```python
from sklearn.preprocessing import StandardScaler # Initialize the scaler scaler = StandardScaler() # Fit and transform the data df_standard = pd.DataFrame(scaler.fit_transform(df), columns=df.columns) print(df_standard.head())
```
